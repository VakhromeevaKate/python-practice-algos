```python
#мемоизация
def climbing_stairs_memo(n, memo=None):
    if memo is None:
        memo = {}
    if n <= 1:
        return 1
    if n in memo:
        return memo[n]
    memo[n] = climbing_stairs_memo(n - 1, memo) + climbing_stairs_memo(n - 2, memo)
    return memo[n]
```

```python
#табуляция
def climbing_stairs_tabulation(n):
    if n <= 1:
        return 1
    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]
```

```python
#оптимизированная табуляция
def climbing_stairs_optimized_tabulation(n):
    if n <= 1:
        return 1
    prev, curr = 1, 1
    for _ in range(2, n + 1):
        prev, curr = curr, prev + curr
    return curr
```

```python
print('test')
for n in [2, 3, 5]:
    print(f"n = {n}")
    print("  memo:     ", climbing_stairs_memo(n))
    print("  tabulation:", climbing_stairs_tabulation(n))
    print("  optimized:", climbing_stairs_optimized_tabulation(n))
```

```python
#мемоизация
def house_robber_memo(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    
    memo = {}
    
    def dp(i):
        if i == 0:
            return nums[0]
        if i == 1:
            return max(nums[0], nums[1])
        if i in memo:
            return memo[i]
        memo[i] = max(dp(i - 1), dp(i - 2) + nums[i])
        return memo[i]
        
    return dp(len(nums) - 1)
```

```python
#табуляция
def house_robber_tabulation(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
        
    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])
    
    for i in range(2, len(nums)):
        dp[i] = max(dp[i - 1], dp[i - 2] + nums[i])
        
    return dp[-1]
```

```python
#оптимизированная табуляция
def house_robber_optimized_tabulation(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
        
    prev2 = nums[0]
    prev1 = max(nums[0], nums[1])
    
    for i in range(2, len(nums)):
        curr = max(prev1, prev2 + nums[i])
        prev2 = prev1
        prev1 = curr
        
    return prev1
```

```python
print('test')
test_cases = [
    [1, 2, 3, 1],
    [2, 7, 9, 3, 1],
    [],
    [5],
    [2, 1]
]

for nums in test_cases:
    print(f"nums = {nums}")
    print("  memo:     ", house_robber_memo(nums))
    print("  tabulation:", house_robber_tabulation(nums))
    print("  optimized:", house_robber_optimized_tabulation(nums))
```

```python
def house_robber_circular(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    def rob_line(arr):
        if not arr:
            return 0
        if len(arr) == 1:
            return arr[0]
        prev2 = arr[0]
        prev1 = max(arr[0], arr[1])
        for i in range(2, len(arr)):
            curr = max(prev1, prev2 + arr[i])
            prev2 = prev1
            prev1 = curr
        return prev1
    return max(rob_line(nums[:-1]), rob_line(nums[1:]))


print(house_robber_circular([2, 3, 2]))
print(house_robber_circular([1, 2, 3, 1]))
```
