```python
#задача 1
def solution1(strs):
    output = ""
    '''
    strs: list of strings
    output: string
    '''
    if not strs:
        return ""
    output = strs[0]
    for s in strs[1:]:
        while not s.startswith(output):
            output = output[:-1]
            if not output:
                return ""
    return output
```

```python
print('test')
print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
```

```python
#задача 2
def solution2(arr, target):
    output = -1
    '''
    arr: int
    target: int
    output: int
    '''

    left, right = 0, len(arr)
    while left < right:
        mid = (left + right) // 2
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid
    output = left
    return output
```

```python
print('test')
print(solution2([1, 2, 3, 6, 9], 9))
print(solution2([1, 2, 3, 6, 9], 5))
print(solution2([1, 2, 3, 6, 9], 0))
print(solution2([1, 2, 3, 6, 9], 110))
```
