### Climbing Stairs
Условие:
Вы поднимаетесь по лестнице из n ступенек. За один шаг можно подняться на 1 или 2 ступеньки. Сколько существует различных способов добраться до вершины?

Примеры:
n = 2 → 2 (1+1, 2)
n = 3 → 3 (1+1+1, 1+2, 2+1)

Подсказка:

Обозначим dp[i] — число способов дойти до ступеньки i.
На ступеньку i можно попасть с i-1 (шаг 1) или с i-2 (шаг 2).
Значит, dp[i] = dp[i-1] + dp[i-2].
База: dp[0] = 1, dp[1] = 1.

Это как числа Фибоначчи, только со сдвигом.
Требование: решить тремя способами — мемоизация, табуляция, O(1) по памяти.
Сложность: O(n) по времени, O(1) по памяти (для оптимального решения).

```python
# МЕМОИЗАЦИЯ             О(n) - по времени, О(n) - по памяти
def climbing_stairs_memo(n, memo=None):
      if memo is None:
        memo = {}  # создаем словарь для хранения значений
      if n <= 2:
        return n
      if n in memo:
        return memo[n]  # используем этот словарь

      memo[n] = climbing_stairs_memo(n-1, memo) + climbing_stairs_memo(n-2, memo)  # заполняем словарь
      return memo[n]

# тесты
assert climbing_stairs_memo(2, None) == 2
assert climbing_stairs_memo(3, None) == 3
assert climbing_stairs_memo(7, None) == 21
```

```python
# или
from functools import lru_cache # Автоматически запоминает результаты вызовов функции

def climbing_stairs_m(n):
  if n <= 2:
    return n
  return climbing_stairs_m(n-1) + climbing_stairs_m(n-2)
```

```python
# ТАБУЛЯЦИЯ               О(n) - по времени, О(n) - по памяти
def climbing_stairs_tabulaion(n):
  if n <= 2:
    return n
  cache = [0]*(n+1) # создаем массив с числами, к.будем заполнять, пока не пойдем до n-ой ступеньки
  cache[1] = 1
  cache[2] = 2
  for i in range(3, n+1):  # с 3й супеньки, а не 2, как fib
    cache[i] = cache[i-1] + cache[i-2]
  return cache[n]

# тесты
assert climbing_stairs_tabulaion(2) == 2
assert climbing_stairs_tabulaion(3) == 3
assert climbing_stairs_tabulaion(7) == 21
```

```python
# ОПТИМИЗИРОВАННАЯ ТАБУЛЯЦИЯ (по памяти)     О(n) - по времени, О(1) - по памяти
def climbing_stairs_optimized_tabulation(n):
  if n <= 1:
    return n
  prev2 = 1
  prev1 = 2
  for i in range(3, n+1):
    curr = prev1 + prev2
    prev1, prev2 = curr, prev1
  return curr

# тесты
assert climbing_stairs_tabulaion(2) == 2
assert climbing_stairs_tabulaion(3) == 3
assert climbing_stairs_tabulaion(7) == 21
```

### House Robber (*)
Условие:
Вы — грабитель, планирующий ограбить дома на улице. В каждом доме i лежит nums[i] денег. Единственное ограничение: нельзя грабить два соседних дома — сработает сигнализация. Найдите максимальную сумму, которую можно украсть.

nums = [1,2,3,1]   → 4 (грабим дома 0 и 2: 1 + 3)
nums = [2,7,9,3,1] → 12 (грабим дома 0, 2, 4: 2 + 9 + 1)

Подсказка:
Обозначим dp[i] — максимальная сумма, которую можно украсть из первых i домов.
На каждом шаге выбор: ограбить дом i (nums[i] + dp[i-2]) или пропустить его (dp[i-1]).
Формула: dp[i] = max(dp[i-1], dp[i-2] + nums[i]).
База: dp[0] = nums[0], dp[1] = max(nums[0], nums[1]).
Оптимизация: можно хранить только две последние переменные → O(1) памяти.
Сложность: O(n) по времени, O(1) по памяти.

На что обратить внимание:
Краевые случаи: пустой массив, один дом, два дома.
** Подумайте, как модифицировать решение, если дома стоят по кругу.

```python
def house_robber_memo(nums, memo=None):
      if memo is None:
        memo = {}
      n = len(nums)
      if n == 0:
        return "введен пустой массив"
      if n == 1:
        memo[1] = nums[0]
        return memo[1]
      if n == 2:
        memo[2] = max(nums[0], nums[1])
        return memo[2]
      if n in memo:
        return memo[n]
      memo[n] = max(house_robber_memo(nums[0:n-2], memo) + nums[n-1], house_robber_memo(nums[0:n-1], memo))
      #print(n, memo)
      return memo[n]

#print(house_robber_memo([1,10,8,3,9,10,11], memo=None))
```

```python
# тесты
assert house_robber_memo([1,2,3,1]) == 4
assert house_robber_memo([2,7,9,3,1]) == 12
assert house_robber_memo([2,10,9,3,10,40]) == 53 # через один [1,3,5]
assert house_robber_memo([]) == "введен пустой массив"
assert house_robber_memo([1,3]) == 3
assert house_robber_memo([1]) == 1
assert house_robber_memo([1,10,8,3,9,10,11]) == 30 # не через один, пропуская [1,4,6]
```

```python
def house_robber_tabulation(nums):
  n = len(nums)
  if n == 0:
    return "введен пустой массив"
  dp = [0]*(n)
  dp[0] = nums[0]
  if n == 1:
    return dp[0]
  dp[1] = max(nums[0], nums[1])
  if n == 2:
    return dp[1]
  for i in range(2, n):
    dp[i] = max(dp[i-1], dp[i-2] + nums[i])
  return dp[i]

#print(house_robber_tabulation([1,10,8,3,9,10,11]))  # 30, [1, 10, 10, 13, 19, 23, 30]
```

```python
# тесты
assert house_robber_tabulation([1,2,3,1]) == 4
assert house_robber_tabulation([2,7,9,3,1]) == 12
assert house_robber_tabulation([2,10,9,3,10,40]) == 53 # через один [1,3,5]
assert house_robber_tabulation([]) == "введен пустой массив"
assert house_robber_tabulation([1,3]) == 3
assert house_robber_tabulation([1]) == 1
assert house_robber_tabulation([1,10,8,3,9,10,11]) == 30 # не через один, пропуская [1,4,6]
```

```python
def house_robber_optimized_tabulation(nums):
  n = len(nums)
  if n == 0:
    return "введен пустой массив"
  if n == 1:
    return nums[0]
  dp = [0]*(n)
  prev2 = nums[0]
  prev1 = max(nums[0], nums[1])
  curr = prev1
  for i in range(2, n):
    curr = max(prev1, prev2 + nums[i])
    prev1, prev2 = curr, prev1
  return curr
```

```python
# тесты
assert house_robber_optimized_tabulation([1,2,3,1]) == 4
assert house_robber_optimized_tabulation([2,7,9,3,1]) == 12
assert house_robber_optimized_tabulation([2,10,9,3,10,40]) == 53 # через один [1,3,5]
assert house_robber_optimized_tabulation([]) == "введен пустой массив"
assert house_robber_optimized_tabulation([1,3]) == 3
assert house_robber_optimized_tabulation([1]) == 1
assert house_robber_optimized_tabulation([1,10,8,3,9,10,11]) == 30 # не через один, пропуская [1,4,6]
```

```python
def house_robber_cell_tab(nums):
  n = len(nums)
  if n == 0:
    return "введен пустой массив"
  if n == 1:
    return nums[0]
  if n == 2:
    return "не получится грабить"
  # для случая: не берем 1й дом
  fircase = 0
  dp = [0]*(n-1)
  dp[0] = nums[1]
  dp[1] = max(nums[2], nums[1])
  for i in range(3, n):
      dp[i-1] = max(dp[i-2], dp[i-3] + nums[i])
  fircase = dp[i-1]
  #print(fircase, dp)

  # для случая: не берем последний дом
  seccase = 0
  dp = [0]*(n-1)
  dp[0] = nums[0]
  dp[1] = max(nums[0], nums[1])
  for i in range(2, n-1):
      dp[i] = max(dp[i-1], dp[i-2] + nums[i])
  seccase = dp[i]
  #print("----")
  #print(seccase, dp)
  return max(fircase, seccase) # берем наиб сумму из 2х случаев
```
