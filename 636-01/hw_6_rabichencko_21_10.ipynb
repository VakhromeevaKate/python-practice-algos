```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, adjusted_rand_score

data = load_wine()
X =data.data
y =data.target
scaler =StandardScaler()
X_scaled = scaler.fit_transform(X)
dbscan= DBSCAN(eps=1.5,min_samples=5)
labels = dbscan.fit_predict(X_scaled)
clusters =set(labels)
print(f"Метки, которые получились {clusters} -1 шум")

if len(clusters - {-1})>=1 and len(clusters)>1:
    sil = silhouette_score(X_scaled, labels)
    print(f"Силуэт качество кластеров {sil:.3f}")
else:
    print("Силуэт: нужен хотя нормальный кластер")
ari= adjusted_rand_score(y, labels)
print(f"Сходство с реальными ответами ARI {ari:.3f}")
```
