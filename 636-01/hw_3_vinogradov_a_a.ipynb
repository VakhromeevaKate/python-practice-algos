### Climbing Stairs
Условие:
Вы поднимаетесь по лестнице из n ступенек. За один шаг можно подняться на 1 или 2 ступеньки. Сколько существует различных способов добраться до вершины?

Примеры:
n = 2 → 2 (1+1, 2)
n = 3 → 3 (1+1+1, 1+2, 2+1)

Подсказка:

Обозначим dp[i] — число способов дойти до ступеньки i.
На ступеньку i можно попасть с i-1 (шаг 1) или с i-2 (шаг 2).
Значит, dp[i] = dp[i-1] + dp[i-2].
База: dp[0] = 1, dp[1] = 1.

Это как числа Фибоначчи, только со сдвигом.
Требование: решить тремя способами — мемоизация, табуляция, O(1) по памяти.
Сложность: O(n) по времени, O(1) по памяти (для оптимального решения).

```python
def climb_stairs_memo(n, memo=None):
    if memo is None:
        memo = {}

    if n == 0 or n == 1:
        return 1

    if n in memo:
        return memo[n]

    memo[n] = climb_stairs_memo(n - 1, memo) + climb_stairs_memo(n - 2, memo)
    return memo[n]


def climb_stairs_tab(n):
    if n == 0 or n == 1:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


def climb_stairs_optimal(n):
    if n == 0 or n == 1:
        return 1

    prev, curr = 1, 1

    for _ in range(2, n + 1):
        prev, curr = curr, prev + curr

    return curr

for n in [0, 1, 2, 3, 5, 10]:
    print(
        f"n = {n}: "
        f"memo = {climb_stairs_memo(n)}, "
        f"tab = {climb_stairs_tab(n)}, "
        f"optimal = {climb_stairs_optimal(n)}"
    )

# Сложность:
# climb_stairs_memo:    время O(n), память O(n)
# climb_stairs_tab:     время O(n), память O(n)
# climb_stairs_optimal: время O(n), память O(1)
```

### House Robber
Условие:
Вы — грабитель, планирующий ограбить дома на улице. В каждом доме i лежит nums[i] денег. Единственное ограничение: нельзя грабить два соседних дома — сработает сигнализация. Найдите максимальную сумму, которую можно украсть.

nums = [1,2,3,1]   → 4 (грабим дома 0 и 2: 1 + 3)
nums = [2,7,9,3,1] → 12 (грабим дома 0, 2, 4: 2 + 9 + 1)

Подсказка:
Обозначим dp[i] — максимальная сумма, которую можно украсть из первых i домов.
На каждом шаге выбор: ограбить дом i (nums[i] + dp[i-2]) или пропустить его (dp[i-1]).
Формула: dp[i] = max(dp[i-1], dp[i-2] + nums[i]).
База: dp[0] = nums[0], dp[1] = max(nums[0], nums[1]).
Оптимизация: можно хранить только две последние переменные → O(1) памяти.
Сложность: O(n) по времени, O(1) по памяти.

На что обратить внимание:
Краевые случаи: пустой массив, один дом, два дома.
** Подумайте, как модифицировать решение, если дома стоят по кругу.

```python
def house_robber(nums):
    """
    nums: список сумм денег в домах.
    Возвращает максимальную сумму, которую можно украсть,
    не грабя два соседних дома.
    """
    prev2 = 0
    prev1 = 0

    for money in nums:
        current = max(prev1, prev2 + money)
        prev2 = prev1
        prev1 = current

    return prev1


print(house_robber([]))
print(house_robber([5]))
print(house_robber([2, 7]))
print(house_robber([1, 2, 3, 1]))
print(house_robber([2, 7, 9, 3, 1]))

def house_robber_circle(nums):
    if not nums:
        return 0

    if len(nums) == 1:
        return nums[0]

    return max(
        house_robber(nums[:-1]),
        house_robber(nums[1:])
    )


print("По кругу:")
print(house_robber_circle([2, 3, 2]))       # 3
print(house_robber_circle([1, 2, 3, 1]))    # 4
print(house_robber_circle([1, 2, 3]))       # 3

# Сложность:
# house_robber:        время O(n), память O(1)
# house_robber_circle: время O(n), память O(n) из-за срезов nums[:-1] и nums[1:]
# (сам алгоритм внутри по дополнительной памяти O(1))
```
