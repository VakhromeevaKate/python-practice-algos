### Climbing Stairs
Условие:
Вы поднимаетесь по лестнице из n ступенек. За один шаг можно подняться на 1 или 2 ступеньки. Сколько существует различных способов добраться до вершины?

Примеры:
n = 2 → 2 (1+1, 2)
n = 3 → 3 (1+1+1, 1+2, 2+1)

Подсказка:

Обозначим dp[i] — число способов дойти до ступеньки i.
На ступеньку i можно попасть с i-1 (шаг 1) или с i-2 (шаг 2).
Значит, dp[i] = dp[i-1] + dp[i-2].
База: dp[0] = 1, dp[1] = 1.

Это как числа Фибоначчи, только со сдвигом.
Требование: решить тремя способами — мемоизация, табуляция, O(1) по памяти.
Сложность: O(n) по времени, O(1) по памяти (для оптимального решения).

```python
def climbing_stairs_memo(n, memo=None):
    # your solution
    if memo == None:
      memo = {}
    if n <= 1:
      return 1
    if n in memo:
      return memo[n]
    memo[n] = climbing_stairs_memo(n - 1, memo) + climbing_stairs_memo(n - 2, memo)
    return memo[n]
```

```python
import time
for n in [3, 10, 20, 30, 35, 38, 50, 75, 756]:
    t = time.time()
    print(climbing_stairs_memo(n))
    print(f"memo({n}): {time.time() - t:.4f} сек")
```

```python
def climbing_stairs_tabulaion(n):
    # your solution
    if n <= 1:
      return 1
    dp = [0] * (n+1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2, n+1):
      dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
```

```python
import time
for n in [3, 10, 20, 30, 35, 38, 50, 75, 756, 1590, 5000, 50000]:
    t = time.time()
    climbing_stairs_tabulaion(n)
    print(f"memo({n}): {time.time() - t:.4f} сек")
```

```python
def climbing_stairs_optimized_tabulation(n):
    # your solution
    if n <= 1:
      return 1
    prev, cur = 1, 1
    for _ in range(2, n+1):
      prev, cur = cur, prev + cur
    return cur
```

```python
import time
for n in [3, 10, 20, 30, 35, 38, 50, 75, 756, 1590, 5000, 50000]:
    t = time.time()
    climbing_stairs_optimized_tabulation(n)
    print(f"memo({n}): {time.time() - t:.4f} сек")
```

### House Robber (*)
Условие:
Вы — грабитель, планирующий ограбить дома на улице. В каждом доме i лежит nums[i] денег. Единственное ограничение: нельзя грабить два соседних дома — сработает сигнализация. Найдите максимальную сумму, которую можно украсть.

nums = [1,2,3,1]   → 4 (грабим дома 0 и 2: 1 + 3)
nums = [2,7,9,3,1] → 12 (грабим дома 0, 2, 4: 2 + 9 + 1)

Подсказка:
Обозначим dp[i] — максимальная сумма, которую можно украсть из первых i домов.
На каждом шаге выбор: ограбить дом i (nums[i] + dp[i-2]) или пропустить его (dp[i-1]).
Формула: dp[i] = max(dp[i-1], dp[i-2] + nums[i]).
База: dp[0] = nums[0], dp[1] = max(nums[0], nums[1]).
Оптимизация: можно хранить только две последние переменные → O(1) памяти.
Сложность: O(n) по времени, O(1) по памяти.

На что обратить внимание:
Краевые случаи: пустой массив, один дом, два дома.
** Подумайте, как модифицировать решение, если дома стоят по кругу.

```python
def house_robber_memo(nums, m=None):
    # your solution
    if m is None:
      m = {}

    if not nums:
      return 0
    if len(nums) == 1:
      return nums[0]
    if len(nums) == 2:
      return max(nums)
    if len(nums) in m:
      return m[len(nums)]

    m[len(nums)] = max(house_robber_memo(nums[:-1], m), house_robber_memo(nums[:-2], m) + nums[-1])

    return m[len(nums)]
```

```python
print(house_robber_memo([1, 3]))
print(house_robber_memo([]))
print(house_robber_memo([1]))
print(house_robber_memo([3, 1]))
print(house_robber_memo([3, 3]), "\n")

print(house_robber_memo([6, 3, 3, 4, 5, 6, 7]))
print(house_robber_memo([2,7,9,3,1]))
print(house_robber_memo([1,2,3,1]))
```

```python
def house_robber_tabulation(nums):
    # your solution
    if not nums:
      return 0
    if len(nums) == 1:
      return nums[0]
    if len(nums) == 2:
      return max(nums)

    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, len(nums)):
      dp[i] = max(dp[i-1], dp[i-2] + nums[i])
    return dp[len(nums)-1]
```

```python
print(house_robber_tabulation([1, 3]))
print(house_robber_tabulation([]))
print(house_robber_tabulation([1]))
print(house_robber_tabulation([3, 1]))
print(house_robber_tabulation([3, 3]), "\n")

print(house_robber_tabulation([6, 3, 3, 4, 5, 6, 7]))
print(house_robber_tabulation([2,7,9,3,1]))
print(house_robber_tabulation([1,2,3,1]))
```

```python
def house_robber_optimized_tabulation(nums):
    # your solution
    if not nums:
      return 0
    if len(nums) == 1:
      return nums[0]
    if len(nums) == 2:
      return max(nums)

    prev, cur = nums[0], max(nums[0], nums[1])
    for i in range(2, len(nums)):
      prev, cur = cur, max(cur, prev + nums[i])
    return cur
```

```python
print(house_robber_optimized_tabulation([1, 3]))
print(house_robber_optimized_tabulation([]))
print(house_robber_optimized_tabulation([1]))
print(house_robber_optimized_tabulation([3, 1]))
print(house_robber_optimized_tabulation([3, 3]), "\n")

print(house_robber_optimized_tabulation([6, 3, 3, 4, 5, 6, 7]))
print(house_robber_optimized_tabulation([2,7,9,3,1]))
print(house_robber_optimized_tabulation([1,2,3,1]))
```
