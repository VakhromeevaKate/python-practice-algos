### Задача 1. Самый длинный общий префикс

Напишите функцию для поиска самого длинного общего префикса в массиве строк.

Пример 1:
strs = ["flower", "flow", "flight"]
output = "fl"

Пример 2:
strs = ["car", "star", "wars"]
output = ""

Ограничения: len(strs) < 10^5, длина каждой строки < 10^3.

```python
def solution1(strs):
    '''
    strs: list of strings
    output: string
    '''
    output = ""
    count = 0     # счетчик кол-ва одинаковых букв

    # Обработка пустого списка и списка из одной строки
    if not strs:
        return ""
    if len(strs) == 1:
        return strs[0]

    shortest_word = strs[0]   #*чтобы использовать длину самого короткого слова (чтобы не было выхода за range)
    for w in strs[1:]:
        if len(w) < len(shortest_word):
            shortest_word = w   #*


    for j in range(len(shortest_word)):  # сначала цикл по буквам в словах
        for i in range(len(strs)):       # потом цикл по словам в массиве
          if strs[0][j] != strs[i][j]:   # сравниваем буквы первого слова со всеми остальными словами и буквами
            return output                # если 1е буквы не совпадают у первых же 2х слов, дальше искать не имеет смысла

          count += 1                     # счетник, сколько раз появилась одна и та же буква в словах

        if count == len(strs):        # проверка, что буква появилась во всех словах массива
          output += strs[i][j]        # добавляем в ответ букву
          count = 0                   # обнуляем счетник буквы

    return output
```

```python
print('-----Tests-----')
print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
print(solution1(["finger", "finge", "fingertd"]))
print(solution1(["f", "finger"]))
print(solution1(["fiiiinger", "finge", "fingertd", "fing"]))
print(solution1(["hinger", "hflow", "hgfhfi"]))
print(solution1(["", "finger"]))
print(solution1([""]))
print(solution1(["а"]))
```

### Сложность решения:

Память - O(1) (при увеличении входа алгоритму памяти гораздо больше не надо)

Время - O(n^2+n) (вложенный цикл + цикл для поиска короткого слова)

### Задача 2. Позиция вставки

Для отсортированного массива уникальных целых чисел найдите индекс, на которой находится число target, если оно найдено в массиве, и индекс, на который его бы следовало поставить, если число не найдено.

Пример 1:
arr = [1, 2, 3, 6, 9], target = 9
output: 4

Пример 2:
arr = [1, 2, 3, 6, 9], target = 5
output: 3

Пример 3:
arr = [1, 2, 3, 6, 9], target = 0
output: 0

Пример 4:
arr = [1, 2, 3, 6, 9], target = 110
output: 5

Ограничения:
0 <= len(arr) <= 10^3
target и arr[i] - целые числа, target <= 10^5 и arr[i] <= 10^5

```python
def solution2(arr, target):
    output = -1
    '''
    arr: int
    target: int
    output: int
    '''
    if len(arr) == 0:  # случай с пустым массивом
      return 0

    for ind, num in enumerate(arr):
      if target == num: # если target найдено в массиве
        return ind
      else:
        if target < num: # если target меньше, то занимает место первого числа, меньше которого
          return ind
        else: output = ind + 1  # если target больше

    return output
```

```python
print('-----Tests-----')
print(solution2([1, 2, 3, 6, 9], 9))
print(solution2([1, 2, 3, 6, 9], 5))
print(solution2([1, 3, 6, 9], 2))
print(solution2([1, 2, 3, 6, 9], 0))
print(solution2([1, 2, 3, 6, 9], 110))
print(solution2([1], 110))
print(solution2([], 110))
```

### Сложность решения:

Память - O(1)

Время - O(n)  1 цикл
