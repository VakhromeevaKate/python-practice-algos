###1. Задание, память O(m) и время O(n · m)

```python
def longest_common_prefix(strs):
    if not strs:
        return ""

    first = strs[0]

    for i in range(len(first)):
        char = first[i]

        for other in strs[1:]:
            if i >= len(other) or other[i] != char:
                return first[:i]

    return first
```

```python
print('-----Tests-----')
print(longest_common_prefix(["flower", "flow", "flight"]))
print(longest_common_prefix(["car", "star", "wars"]))
print(longest_common_prefix(["test", "test", "test"]))
print(longest_common_prefix(["carrot", "cat", "coconut"]))
```

### 2. Задание, память O(1) <-- ну вроде так, но я не уверена :(((, а время O(log n)

```python
def search_insert(arr, target):
    left = 0
    right = len(arr)

    while left < right:
        mid = left + (right - left) // 2
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid

    return left
```

```python
print('-----Tests-----')
print(search_insert([1, 2, 3, 6, 9], 9))
print(search_insert([1, 2, 3, 6, 9], 5))
print(search_insert([1, 2, 3, 6, 9], 0))
print(search_insert([1, 2, 3, 6, 9], 110))
```
