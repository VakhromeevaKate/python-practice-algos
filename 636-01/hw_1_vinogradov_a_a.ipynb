```python
# Задача 1. Самый длинный общий префикс

def solution1(strs):
    output = ""
    if not strs:
        return output
    min_len = min(len(s) for s in strs)

    for i in range(min_len):
        current_char = strs[0][i]

        for s in strs[1:]:
            if s[i] != current_char:
                return output

        output += current_char

    return output
print('-----Tests-----')
print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
```

```python
Сложность решения:
Память - O(1)
Время - O(n * m), где:
n — количество строк;
m — длина самой короткой строки.
```

```python
# Задача 2. Позиция вставки

def solution2(arr, target):
    left = 0
    right = len(arr)
    while left < right:
        middle = (left + right) // 2

        if arr[middle] < target:
            left = middle + 1
        else:
            right = middle

    return left
print('-----Tests-----')
print(solution2([1, 2, 3, 6, 9], 9))
print(solution2([1, 2, 3, 6, 9], 5))
print(solution2([1, 2, 3, 6, 9], 0))
print(solution2([1, 2, 3, 6, 9], 110))
```

```python
Сложность решения:
Память - O(1)
Время - O(log n), где n — длина массива.
```
