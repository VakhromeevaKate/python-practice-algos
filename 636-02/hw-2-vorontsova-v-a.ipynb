### 1. Задача «Банкомат»

```python
def solution1(amount, limits):
    denominations = [100, 200, 500, 1000, 2000, 5000]
    result = [0, 0, 0, 0, 0, 0]
    
    # Идем от самых крупных купюр к мелким
    for i in range(5, -1, -1):
        if amount <= 0:
            break
        nominal = denominations[i]
        # Сколько купюр можем взять (по сумме и по лимиту)
        count = min(amount // nominal, limits[i])
        result[i] = count
        amount -= count * nominal
        
    if amount > 0:
        return -1
    return result
```

```python
print('-----Tests ATM-----')
print(solution1(700, [10, 10, 10, 10, 10, 10]))
print(solution1(300, [0, 0, 0, 0, 0, 0]))
print(solution1(350, [10, 0, 10, 10, 10, 10]))
print(solution1(0, [0, 0, 0, 0, 0, 0]))
```

### 2. Задача «Подарочная карта»

```python
def solution2(prices, target):
    seen = {}
    for i in range(len(prices)):
        price = prices[i]
        needed = target - price
        if needed in seen:
            return [seen[needed], i]
        seen[price] = i
    return []
```

```python
print('-----Tests Gift Card-----')
print(solution2([2, 7, 11, 15], 9))
print(solution2([3, 3], 6))
```
