```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_squared_error
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

california_data = fetch_california_housing(as_frame=True)
df = california_data.frame
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(numeric_only=True), annot = True)
plt.title("Correlation Matrix")
plt.show()
```

```python
num_cols = ['MedInc', 'HouseAge', 'AveRooms', 'AveBedrms', 'Population', 'AveOccup', 'Latitude', 'Longitude']

y = df['MedHouseVal']
x = df.drop('MedHouseVal', axis=1)
```

```python
print('Shape:', x.shape)
```

```python
x.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled[0]
```

```python
models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(),
    'Lasso': Lasso(),
    'RandomForestRegressor': RandomForestRegressor(),
}

default_acc = {}
for name, model in models.items():
    model.fit(x_train_scaled, y_train)
    default_acc[name] = mean_squared_error(y_test, model.predict(x_test_scaled))
    print(name, '->', round(default_acc[name], 4))
```

## Подбор гиперпараметров (GridSearchCV)

```python
param_grids = {
    'Ridge': (Ridge(), {'alpha': [0.01, 0.1, 1, 10, 100]}),
    'Lasso': (Lasso(max_iter=10000), {'alpha': [0.0001, 0.001, 0.01, 0.1, 1]}),
    'RandomForestRegressor': (
        RandomForestRegressor(random_state=42),
        {'n_estimators': [100, 200], 'max_depth': [None, 20], 'min_samples_leaf': [1, 2]},
    ),
}

best_models = {}
tuned_acc = {}
for name, (model, grid) in param_grids.items():
    gs = GridSearchCV(model, grid, cv=3, scoring='neg_mean_squared_error', n_jobs=-1)
    gs.fit(x_train_scaled, y_train)
    best_models[name] = gs.best_estimator_
    tuned_acc[name] = mean_squared_error(y_test, gs.best_estimator_.predict(x_test_scaled))
    print(name, '->', gs.best_params_)
    print('   CV MSE:', round(-gs.best_score_, 4), '| test MSE:', round(tuned_acc[name], 4))
```

### Сравнение моделей до и после подбора гиперпараметров

```python
tuned_all = {'Linear Regression': default_acc['Linear Regression'], **tuned_acc}

results = pd.DataFrame({
    'MSE (по умолчанию)': default_acc,
    'MSE (после подбора)': tuned_all,
}).round(4)
print(results)

results.plot(kind='bar', figsize=(10, 5))
plt.ylabel('MSE на тестовой выборке')
plt.title('Сравнение моделей')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
best_name = min(tuned_all, key=tuned_all.get)
print('Лучшая модель:', best_name)
print('MSE:', round(tuned_all[best_name], 4))
print('RMSE:', round(tuned_all[best_name] ** 0.5, 4), '(в единицах целевой переменной, т.е. сотнях тысяч $)')
```
