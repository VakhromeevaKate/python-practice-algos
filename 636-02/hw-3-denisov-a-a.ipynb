```python
#1
#Мемоизация
def climbing_stairs_memo(n):
    if n < 0:
        return 0
    memo = {}
    def dp(i):
        if i <= 1:
            return 1
        if i in memo:
            return memo[i]
        memo[i] = dp(i - 1) + dp(i - 2)
        return memo[i]
    return dp(n)

#Табуляция
def climbing_stairs_tabulaion(n):
    if n < 0:
        return 0
    if n <= 1:
        return 1
    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]

#Оптимизированная табуляция
def climbing_stairs_optimized_tabulation(n):
    if n < 0:
        return 0
    if n <= 1:
        return 1
    prev2, prev1 = 1, 1
    for _ in range(2, n + 1):
        curr = prev1 + prev2
        prev2, prev1 = prev1, curr
    return prev1
```

```python
for n in [0, 1, 2, 3, 4, 5, 100]:
    print(f"n={n}: memo={climbing_stairs_memo(n)}, "
          f"tab={climbing_stairs_tabulaion(n)}, "
          f"opt={climbing_stairs_optimized_tabulation(n)}")
```

Мемоизация:

*   Время: O(n) — каждый подзадача вычисляется один раз
*   Память: O(n) — словарь для хранения результатов + стек рекурсии глубиной n

Табуляция:
*   Время: O(n) — один проход от 2 до n
*   Память: O(n) — массив dp размером n+1

Оптимизированная табуляция:
*   Время: O(n) — один проход от 2 до n
*   Память: O(1) — храним только две переменные


```python

def house_robber_memo(nums):
    if not nums:
        return 0
    from functools import lru_cache
    @lru_cache(maxsize=None)
    def dp(i):
        if i == 0:
            return nums[0]
        if i == 1:
            return max(nums[0], nums[1])
        return max(dp(i - 1), dp(i - 2) + nums[i])
    return dp(len(nums) - 1)

def house_robber_tabulation(nums):
    if not nums:
        return 0
    n = len(nums)
    if n == 1:
        return nums[0]
    dp = [0] * n
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])
    for i in range(2, n):
        dp[i] = max(
            dp[i - 1],
            dp[i - 2] + nums[i]
        )
    return dp[-1]

def house_robber_optimized_tabulation(nums):
    if not nums:
        return 0
    n = len(nums)
    if n == 1:
        return nums[0]
    prev2 = nums[0]
    prev1 = max(nums[0], nums[1])
    for i in range(2, n):
        current = max(prev1, prev2 + nums[i])
        prev2, prev1 = prev1, current
    return prev1
```

```python
print(house_robber_memo([1, 2, 3, 1]))
print(house_robber_tabulation([2, 7, 9, 3, 1]))
print(house_robber_optimized_tabulation([2, 7, 9, 3, 1]))
```

```python
#краевые случаи
print(house_robber_optimized_tabulation([]))
print(house_robber_optimized_tabulation([5]))
print(house_robber_optimized_tabulation([3, 1]))
```

```python
#для домов по кругу
def house_robber_circ(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    def rob_line(line):
        if not line:
            return 0
        if len(line) == 1:
            return line[0]
        prev2 = line[0]
        prev1 = max(line[0], line[1])
        for i in range(2, len(line)):
            current = max(prev1, prev2 + line[i])
            prev2, prev1 = prev1, current
        return prev1
    return max(
        rob_line(nums[:-1]),
        rob_line(nums[1:])
    )
```

```python
print(house_robber_circ([2, 3, 2]))
print(house_robber_circ([2, 7, 9, 3, 1]))
```
