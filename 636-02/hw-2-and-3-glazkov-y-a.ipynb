```python
def Tab(n):
    if n <= 1:
        return 1
    dp = [0]*(n+1)
    dp[0] =1
    dp[1] = 1
    for i in range(2,n+1):
        dp[i] =dp[i-1]+dp[i-2]

    return dp[n]
```

```python
def Opt(n):
    if n <=1:
        return 1
    count1=1
    count2=1

    for i in range(2,n+1):
        s =count1+count2
        count2 =count1
        count2 = s
    return count1
```

```python
def Memoiz(n):
    m = {0:1, 1:1}

    def f(b):
        if b in m:
            return m[b]
        m[b] = f(b-1)+f(b-2)
        return m[b]

    return f(n)
```

```python
#2 g
def grab(a):
    if not a:
        return 0
    if len(a) == 1:
        return a[0]

    max2 =a[0]
    max1 = max(a[0],a[1])
    for i in range(2,len(a)):
        s =max(max1, max2+a[i])
        max2 = max1
        max1 =s

    return max1
```

```python
grab([1, 2, 3, 1])
```
