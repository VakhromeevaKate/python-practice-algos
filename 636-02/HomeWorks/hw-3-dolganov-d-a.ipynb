### Climbing Stairs
Условие:
Вы поднимаетесь по лестнице из n ступенек. За один шаг можно подняться на 1 или 2 ступеньки. Сколько существует различных способов добраться до вершины?

Примеры:
n = 2 → 2 (1+1, 2)
n = 3 → 3 (1+1+1, 1+2, 2+1)

Подсказка:

Обозначим dp[i] — число способов дойти до ступеньки i.
На ступеньку i можно попасть с i-1 (шаг 1) или с i-2 (шаг 2).
Значит, dp[i] = dp[i-1] + dp[i-2].
База: dp[0] = 1, dp[1] = 1.

Это как числа Фибоначчи, только со сдвигом.
Требование: решить тремя способами — мемоизация, табуляция, O(1) по памяти.
Сложность: O(n) по времени, O(1) по памяти (для оптимального решения).

```python
def climbing_stairs_memo(n):
    # your solution
    memo = {}

    def climb(step):
        if step <= 1:
            return 1

        if step in memo:
            return memo[step]

        memo[step] = climb(step - 1) + climb(step - 2)
        return memo[step]

    return climb(n)
```

```python
print(climbing_stairs_memo(2))
print(climbing_stairs_memo(3))
print(climbing_stairs_memo(5))
```

```python
def climbing_stairs_tabulaion(n):
    # your solution
    if n <= 1:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
```

```python
print(climbing_stairs_tabulaion(2))
print(climbing_stairs_tabulaion(3))
print(climbing_stairs_tabulaion(5))
```

```python
def climbing_stairs_optimized_tabulation(n):
    # your solution
    if n <= 1:
        return 1

    prev = 1
    curr = 1

    for _ in range(2, n + 1):
        prev, curr = curr, prev + curr

    return curr
```

```python
print(climbing_stairs_optimized_tabulation(2))
print(climbing_stairs_optimized_tabulation(3))
print(climbing_stairs_optimized_tabulation(5))
print(climbing_stairs_optimized_tabulation(10))
```

### House Robber (*)
Условие:
Вы — грабитель, планирующий ограбить дома на улице. В каждом доме i лежит nums[i] денег. Единственное ограничение: нельзя грабить два соседних дома — сработает сигнализация. Найдите максимальную сумму, которую можно украсть.

nums = [1,2,3,1]   → 4 (грабим дома 0 и 2: 1 + 3)
nums = [2,7,9,3,1] → 12 (грабим дома 0, 2, 4: 2 + 9 + 1)

Подсказка:
Обозначим dp[i] — максимальная сумма, которую можно украсть из первых i домов.
На каждом шаге выбор: ограбить дом i (nums[i] + dp[i-2]) или пропустить его (dp[i-1]).
Формула: dp[i] = max(dp[i-1], dp[i-2] + nums[i]).
База: dp[0] = nums[0], dp[1] = max(nums[0], nums[1]).
Оптимизация: можно хранить только две последние переменные → O(1) памяти.
Сложность: O(n) по времени, O(1) по памяти.

На что обратить внимание:
Краевые случаи: пустой массив, один дом, два дома.
** Подумайте, как модифицировать решение, если дома стоят по кругу.

```python
def house_robber_memo(nums):
    # your solution
    memo = {}

    def rob(i):
        if i >= len(nums):
            return 0

        if i in memo:
            return memo[i]

        take = nums[i] + rob(i + 2)
        skip = rob(i + 1)

        memo[i] = max(take, skip)
        return memo[i]

    return rob(0)
```

```python
print(house_robber_memo([1, 2, 3, 1]))
print(house_robber_memo([2, 7, 9, 3, 1]))
print(house_robber_memo([]))
print(house_robber_memo([5]))
```

```python
def house_robber_tabulation(nums):
    # your solution
    if not nums:
        return 0

    if len(nums) == 1:
        return nums[0]

    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, len(nums)):
        dp[i] = max(dp[i - 1], dp[i - 2] + nums[i])

    return dp[-1]
```

```python
print(house_robber_tabulation([1, 2, 3, 1]))
print(house_robber_tabulation([2, 7, 9, 3, 1]))
print(house_robber_tabulation([]))
print(house_robber_tabulation([5]))
print(house_robber_tabulation([2, 10]))
```

```python
def house_robber_optimized_tabulation(nums):
    # your solution
    prev = 0
    curr = 0

    for money in nums:
        prev, curr = curr, max(curr, prev + money)

    return curr
```

```python
print(house_robber_optimized_tabulation([1, 2, 3, 1]))
print(house_robber_optimized_tabulation([2, 7, 9, 3, 1]))
print(house_robber_optimized_tabulation([]))
print(house_robber_optimized_tabulation([5]))
print(house_robber_optimized_tabulation([2, 10]))
```
