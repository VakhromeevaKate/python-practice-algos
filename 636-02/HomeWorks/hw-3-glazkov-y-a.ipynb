Задача 1 - Climbing Stars

```python
def climbStairs_memo(n, memo=None):
    """Решение с мемоизацией (Top-Down)"""
    if n <= 0: return 0
    if n <= 2: return n
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    memo[n] = climbStairs_memo(n - 1, memo) + climbStairs_memo(n - 2, memo)
    return memo[n]

def climbStairs_tab(n):
    """Решение с табуляцией (Bottom-Up)"""
    if n <= 0: return 0
    if n <= 2: return n
    dp = [0] * (n + 1)
    dp[1], dp[2] = 1, 2
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]

def climbStairs_optimal(n):
    """Оптимальное решение с O(1) памяти"""
    if n <= 0: return 0
    if n <= 2: return n
    prev, curr = 1, 2
    for _ in range(3, n + 1):
        prev, curr = curr, prev + curr
    return curr

# Тесты
print("Climbing Stairs Tests:")
test_cases = [0, 1, 2, 3, 5, 45] # 45 проверит, что оптимальное решение не падает по времени
for n in test_cases:
    res_m = climbStairs_memo(n)
    res_t = climbStairs_tab(n)
    res_o = climbStairs_optimal(n)
    print(f"n={n:2d} | Memo: {res_m:3d} | Tab: {res_t:3d} | Opt: {res_o:3d} | Match: {res_m == res_t == res_o}")
```

**Сложность решения (Climbing Stairs):**
- **Мемоизация:** Время - $O(n)$, Память - $O(n)$ (словарь + стек вызовов).
- **Табуляция:** Время - $O(n)$, Память - $O(n)$ (массив `dp`).
- **Оптимальное:** Время - $O(n)$, Память - $O(1)$ (хранятся только две последние переменные). Это финальное ожидаемое решение.

```python
def rob(nums):
    """Оптимальное решение для линейной улицы, O(1) памяти"""
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]

    prev2 = 0  # макс. сумма до дома i-2
    prev1 = 0  # макс. сумма до дома i-1

    for num in nums:
        # текущий максимум
        curr = max(prev1, prev2 + num)
        prev2 = prev1
        prev1 = curr

    return prev1

def rob_circular(nums):
    """БОНУС: Модификация для круговой улицы (дома 0 и n-1 соседние)"""
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    # либо грабим с 0 до n-2, либо с 1 до n-1. Максимум из двух вариантов.
    return max(rob(nums[:-1]), rob(nums[1:]))

# Тесты
print("\nHouse Robber Tests:")
# Стандартные тесты
print(f"[1,2,3,1] -> {rob([1,2,3,1])} (Ожидается: 4)")
print(f"[2,7,9,3,1] -> {rob([2,7,9,3,1])} (Ожидается: 12)")

# Краевые условия
print(f"[] (пустой) -> {rob([])} (Ожидается: 0)")
print(f"[5] (один дом) -> {rob([5])} (Ожидается: 5)")
print(f"[2, 1] (два дома) -> {rob([2, 1])} (Ожидается: 2)")

# Бонусный тест (круговая улица)
print(f"\nCircular [2,3,2] -> {rob_circular([2,3,2])} (Ожидается: 3, нельзя брать первый и последний)")
print(f"Circular [1,2,3,1] -> {rob_circular([1,2,3,1])} (Ожидается: 4)")
```
