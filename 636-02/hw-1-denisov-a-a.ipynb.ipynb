```python
#1
def solution1(strs):
    output = ""
    if not strs:
        return output
    for i in range(len(strs[0])):
        char = strs[0][i]
        for j in range(1, len(strs)):
            if i == len(strs[j]) or strs[j][i] != char:
                output = strs[0][:i]
                return output
    output = strs[0]
    return output


print('Tests')
print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
```


Сложность решения:
*   Память - O(1)
*   Время - O(N * M) или O(S), где N — количество строк, M — длина самой короткой строки, а S — суммарное количество символов,в худшем случае алгоритм проверит все символы

```python
#2
def solution2(arr, target):
    output = -1
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            output = mid
            return output
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    output = left
    return output


print('-----Tests-----')
print(solution2([1, 2, 3, 6, 9], 9))
print(solution2([1, 2, 3, 6, 9], 5))
print(solution2([1, 2, 3, 6, 9], 0))
print(solution2([1, 2, 3, 6, 9], 110))
```

Сложность решения:
*   Память - O(1)
*   Время - O(log N), где N — длина массива arr. На каждом шаге бинарного поиска область поиска сокращается вдвое
