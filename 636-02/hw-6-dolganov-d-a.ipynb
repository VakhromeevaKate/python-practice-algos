1. Запустите `DBSCAN` на датасете Wine (с параметрами `eps=1.5, min_samples=5`).
2. Посчитайте `silhouette_score` и `adjusted_rand_score`.
3. Сравните с KMeans. Какой алгоритм лучше справился с поиском истинных сортов? Почему?


```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.cluster import DBSCAN, KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.preprocessing import StandardScaler
```

```python
wine = load_wine()

X = wine.data
y = wine.target

print("Размер данных:", X.shape)
print("Количество классов:", len(np.unique(y)))
print("Названия классов:", wine.target_names)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

```python
dbscan = DBSCAN(eps=1.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)

print("Метки кластеров:", np.unique(dbscan_labels))
print("Количество кластеров:", len(set(dbscan_labels) - {-1}))
print("Количество шумовых точек:", np.sum(dbscan_labels == -1))
```

```python
ari_dbscan = adjusted_rand_score(y, dbscan_labels)

if len(np.unique(dbscan_labels)) > 1:
    silhouette_dbscan = silhouette_score(X_scaled, dbscan_labels)
else:
    silhouette_dbscan = None

print("DBSCAN Adjusted Rand Score:", ari_dbscan)
print("DBSCAN Silhouette Score:", silhouette_dbscan)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

print("Метки кластеров:", np.unique(kmeans_labels))
print("Количество кластеров:", len(np.unique(kmeans_labels)))
```

```python
silhouette_kmeans = silhouette_score(X_scaled, kmeans_labels)
ari_kmeans = adjusted_rand_score(y, kmeans_labels)

print("KMeans Silhouette Score:", silhouette_kmeans)
print("KMeans Adjusted Rand Score:", ari_kmeans)
```

KMeans справился лучше: ARI = 0.897, Silhouette Score = 0.285
DBSCAN не нашёл кластеры и отнёс все точки к шуму
