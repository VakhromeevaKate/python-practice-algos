### Climbing Stairs

```python
def climbing_stairs_memo(n):
    memo = {}
    def helper(k):
        if k <= 1:
            return 1
        if k in memo:
            return memo[k]
        memo[k] = helper(k - 1) + helper(k - 2)
        return memo[k]
    return helper(n)
```

```python
def climbing_stairs_tabulaion(n):
    if n <= 1:
        return 1
    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]
```

```python
def climbing_stairs_optimized_tabulation(n):
    if n <= 1:
        return 1
    prev2 = 1
    prev1 = 1
    current = 0
    for _ in range(2, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current
    return current
```

### House Robber (*)

```python
def house_robber_memo(nums):
    if not nums:
        return 0
    memo = {}
    def helper(i):
        if i == 0:
            return nums[0]
        if i < 0:
            return 0
        if i in memo:
            return memo[i]
        memo[i] = max(helper(i - 1), helper(i - 2) + nums[i])
        return memo[i]
    return helper(len(nums) - 1)
```

```python
def house_robber_tabulation(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])
    for i in range(2, len(nums)):
        dp[i] = max(dp[i - 1], dp[i - 2] + nums[i])
    return dp[-1]
```

```python
def house_robber_optimized_tabulation(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    prev2 = nums[0]
    prev1 = max(nums[0], nums[1])
    current = prev1
    for i in range(2, len(nums)):
        current = max(prev1, prev2 + nums[i])
        prev2 = prev1
        prev1 = current
    return current
```
