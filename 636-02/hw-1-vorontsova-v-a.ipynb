### Задача 1. Самый длинный общий префикс

```python
def solution1(strs):
    if not strs:
        return ""
    
    first_word = strs[0]
    output = ""
    
    for i in range(len(first_word)):
        char = first_word[i]
        for word in strs:
            if i >= len(word) or word[i] != char:
                return output
        output += char
        
    return output
```

```python
print('-----Tests-----')
print(solution1(["flower", "flow", "flight"])) 
print(solution1(["car", "star", "wars"]))
```

### Сложность решения:

Память - O(1)
Время - O(N * M)
