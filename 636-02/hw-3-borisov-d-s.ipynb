### Climbing Stairs
Условие:
Вы поднимаетесь по лестнице из n ступенек. За один шаг можно подняться на 1 или 2 ступеньки. Сколько существует различных способов добраться до вершины?

Примеры:
n = 2 → 2 (1+1, 2)
n = 3 → 3 (1+1+1, 1+2, 2+1)

Подсказка:

Обозначим dp[i] — число способов дойти до ступеньки i.
На ступеньку i можно попасть с i-1 (шаг 1) или с i-2 (шаг 2).
Значит, dp[i] = dp[i-1] + dp[i-2].
База: dp[0] = 1, dp[1] = 1.

Это как числа Фибоначчи, только со сдвигом.
Требование: решить тремя способами — мемоизация, табуляция, O(1) по памяти.
Сложность: O(n) по времени, O(1) по памяти (для оптимального решения).

```python
def climbing_stairs_memo(n):
    # your solution
    if type(n) is not int:
        return 0
    if n is None or n < 0:
        return 0
    memo = {}

    def dp(i):
        if i <= 1:
            return 1
        if i in memo:
            return memo[i]
        memo[i] = dp(i - 1) + dp(i - 2)
        return memo[i]

    return dp(n)
print(climbing_stairs_memo(3))
print(climbing_stairs_memo(-2)) # Отрицательное число ступенек
print(climbing_stairs_memo('STR')) # Передана строка вместо числа
```

```python
def climbing_stairs_tabulaion(n):
    # your solution
    if type(n) is not int:
        return 0
    if n is None or n < 0:
        return 0
    if n <= 1:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
print(climbing_stairs_tabulaion(3))
print(climbing_stairs_tabulaion(-2)) # Отрицательное число ступенек
print(climbing_stairs_tabulaion('STR')) # Передана строка вместо числа
```

```python
def climbing_stairs_optimized_tabulation(n):
    # your solution
    if type(n) is not int:
        return 0
    if n is None or n < 0:
        return 0
    if n <= 1:
        return 1

    prev2 = 1  # dp[i-2], начиная с dp[0]
    prev1 = 1  # dp[i-1], начиная с dp[1]

    for i in range(2, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current

    return prev1
print(climbing_stairs_optimized_tabulation(3))
print(climbing_stairs_optimized_tabulation(-2)) # Отрицательное число ступенек
print(climbing_stairs_optimized_tabulation('STR')) # Передана строка вместо числа
```

### House Robber (*)
Условие:
Вы — грабитель, планирующий ограбить дома на улице. В каждом доме i лежит nums[i] денег. Единственное ограничение: нельзя грабить два соседних дома — сработает сигнализация. Найдите максимальную сумму, которую можно украсть.

nums = [1,2,3,1]   → 4 (грабим дома 0 и 2: 1 + 3)
nums = [2,7,9,3,1] → 12 (грабим дома 0, 2, 4: 2 + 9 + 1)

Подсказка:
Обозначим dp[i] — максимальная сумма, которую можно украсть из первых i домов.
На каждом шаге выбор: ограбить дом i (nums[i] + dp[i-2]) или пропустить его (dp[i-1]).
Формула: dp[i] = max(dp[i-1], dp[i-2] + nums[i]).
База: dp[0] = nums[0], dp[1] = max(nums[0], nums[1]).
Оптимизация: можно хранить только две последние переменные → O(1) памяти.
Сложность: O(n) по времени, O(1) по памяти.

На что обратить внимание:
Краевые случаи: пустой массив, один дом, два дома.
** Подумайте, как модифицировать решение, если дома стоят по кругу.

```python
def house_robber_memo(nums=None):
    # your solution
    if not nums:
        return 0
    if type(nums) is not list:
        if type(nums) is int:
            nums = [nums]
        else:
            return 0

    memo = {}

    def dp(i):
        if i < 0:
            return 0
        if i == 0:
            return nums[0]
        if i in memo:
            return memo[i]
        memo[i] = max(dp(i - 1), dp(i - 2) + nums[i])
        return memo[i]

    return dp(len(nums) - 1)

print(house_robber_memo([43, 24, 25]))
print(house_robber_memo()) # Нет массива
print(house_robber_memo("STR")) # Введён не массив
print(house_robber_memo(2)) # Введён не массив, но число (создадим массив с 1-м значением)
print(house_robber_memo([])) # Пустой массив
print(house_robber_memo([43])) # 1 дом спокойно грабим
print(house_robber_memo([43, 25])) # Максимальное значение из 2-х домов (Оба грабить нельзя)
```

```python
def house_robber_tabulation(nums=None):
    # your solution
    if not nums:
        return 0
    if type(nums) is not list:
        if type(nums) is int:
            nums = [nums]
        else:
            return 0
    if len(nums) == 1:
        return nums[0]

    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, len(nums)):
        dp[i] = max(dp[i - 1], dp[i - 2] + nums[i])

    return dp[-1]

print(house_robber_tabulation([43, 24, 25]))
print(house_robber_tabulation()) # Нет массива
print(house_robber_tabulation("STR")) # Введён не массив
print(house_robber_tabulation(2)) # Введён не массив, но число (создадим массив с 1-м значением)
print(house_robber_tabulation([])) # Пустой массив
print(house_robber_tabulation([43])) # 1 дом спокойно грабим
print(house_robber_tabulation([43, 25])) # Максимальное значение из 2-х домов (Оба грабить нельзя)
```

```python
def house_robber_optimized_tabulation(nums=None):
    # your solution
    if not nums:
        return 0
    if type(nums) is not list:
        if type(nums) is int:
            nums = [nums]
        else:
            return 0
    if len(nums) == 1:
        return nums[0]

    prev2 = nums[0]                    # dp[i-2]
    prev1 = max(nums[0], nums[1])      # dp[i-1]

    for i in range(2, len(nums)):
        current = max(prev1, prev2 + nums[i])
        prev2 = prev1
        prev1 = current

    return prev1

print(house_robber_optimized_tabulation([43, 24, 25]))
print(house_robber_optimized_tabulation()) # Нет массива
print(house_robber_optimized_tabulation("STR")) # Введён не массив
print(house_robber_optimized_tabulation(2)) # Введён не массив, но число (создадим массив с 1-м значением)
print(house_robber_optimized_tabulation([])) # Пустой массив
print(house_robber_optimized_tabulation([43])) # 1 дом спокойно грабим
print(house_robber_optimized_tabulation([43, 25])) # Максимальное значение из 2-х домов (Оба грабить нельзя)
```

```python
#-------------------- Вариант с домами по кругу-----------------------
def house_robber_circular(nums=None):
    if type(nums) is not list:
        if type(nums) is int:
            nums = [nums]
        else:
            return 0
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    return max(house_robber_memo(nums[:-1]), house_robber_memo(nums[1:]))

print(house_robber_circular([5, 3, 2, 3, 5]))
print(house_robber_circular())
print(house_robber_circular([]))
print(house_robber_circular(6))
print(house_robber_circular("STR"))
```
