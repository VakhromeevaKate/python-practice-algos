### Задача 1. Самый длинный общий префикс

```python
def solution1(strs):
    output = ""
    '''
    strs: list of strings
    output: string
    '''
    if not strs:
        return output

    prefix = strs[0]
    for s in strs[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if prefix == "":
                return output
    output = prefix

    return output
```

```python
print('-----Tests-----')
print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
```

### Задача 2. Позиция вставки

```python
def solution2(arr, target):
    output = -1
    '''
    arr: list of int
    target: int
    output: int
    '''

    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    output = left

    return output
```

```python
print('-----Tests-----')
print(solution2([1, 2, 3, 6, 9], 9))
print(solution2([1, 2, 3, 6, 9], 5))
print(solution2([1, 2, 3, 6, 9], 0))
print(solution2([1, 2, 3, 6, 9], 110))
```
